```
rdd = sc.textFile("FileStore/tables/custs_without_header.csv")
```

```
rdd.first()
```

```
rdd.getNumPartitions()
```

```
rdd2=rdd.map(lambda rec: rec.split(","))
```

```
rdd2.first()
```

```
rdd3=rdd2.map(lambda lst:(lst[1],lst[2],lst[3],lst[4]))
```

```
rdd3.first()
```

```
rdd4 = rdd3.filter(lambda tu:tu[0].startswith("S"))
```

```
rdd4.first()
```

```
rdd4.take(5)
```

```
rdd5 = rdd4.filter(lambda tu:int(tu[2]) < 45)
```

```
rdd5.first()
```

```
rdd5.count()
```

```
for i in rdd.collect():
  print(i)
```

```
rdd6.first()
rdd6.count()
```

```
rdd7 = rdd2.map(lambda tu : (tu[1], tu[2], tu[1] + ' ' + tu[2]))
rdd7.first()
```

```
rdd8 = rdd2.filter(lambda tu : tu[2].endswith("s"))
rdd8.first()
```

```
rdd9 = rdd2.filter(lambda tu: int(tu[3]) == 45 and tu[4] == 'Pilot')
rdd9.first()
```
